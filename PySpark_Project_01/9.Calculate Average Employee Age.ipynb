```
schema = StructType([
    StructField("Emp_no",IntegerType(),True),
    StructField ("Emp_name",StringType(),True),
    StructField ("Salary",IntegerType(),True),
    StructField ("Age",IntegerType(),True),
    StructField ("Department",StringType(),True)
])
```

```
df = spark.createDataFrame(data,schema)
df.show(5)
```

```
# Display schema of DataFrame
df.printSchema()
```

```
#  Create a temporary view
df.createOrReplaceTempView("temp_tbl")
```

```
# Execute an SQL query
spark.sql("SELECT * FROM temp_tbl").count()
```

```
spark.sql("SELECT * FROM temp_tbl").show()
```

```
# Calculate Average Salary by Department
spark.sql(""" SELECT 
            Department,
            round(avg(Salary),0) as Avg_Salary 
          FROM temp_tbl
          GROUP BY 1 ORDER BY 2 DESC
          """).show()

```

```
df.groupBy("Department").agg(round(avg("Salary")).alias("Avg_Salary")).show()
```

```
#Add 10% Bonus to Salaries

new_temp_tbl = spark.sql("""
                        SELECT 
                            *,
                            Salary,
                            (Salary * 10/100) as Bonus,
                            Salary + (Salary * 10/100) as New_salary
                        FROM temp_tbl
          """)
```

```
spark.sql ("SELECT * FROM temp_tbl").show(5)
```

```
new_temp_tbl.createOrReplaceTempView("updated_temp_tbl")
spark.sql("SELECT * FROM updated_temp_tbl").show(5)
```

```
df=df.withColumn("Bonus",col("Salary")*10/100).withColumn("New_salary",col("Salary")+ col("Salary")*10/100)
df.show()
```

```
# Find Maximum Salary by Age
spark.sql("""
        SELECT 
            Age,
            Max(Salary) as Max_salary_by_age
        FROM updated_temp_tbl
        GROUP BY 1 ORDER BY 2 DESC
          """).show()
```

```
df.groupby("Age")\
    .agg(max("Salary").alias("Max_salary_by_age"))\
    .orderBy("Max_salary_by_age",ascending = False)\
    .show()
```
