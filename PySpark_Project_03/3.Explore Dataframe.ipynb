### 1.Import and create sparksession

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,lit,sum,count,min,max,count,when,udf,rank
```

```
import pandas as pd 
import plotly.express as px
```

```
spark = SparkSession.builder.appName("offline_paid").getOrCreate()
```

### 2.Read CSV File

```
df = spark.read.table("offline_paid")
display(df.limit(5))
```

### 3.Explore Dataframe

```
df.printSchema()
```

```
schema_data = [ (
    field.name,
    field.dataType.simpleString(),
    field.nullable
    )for field in df.schema.fields]
schema_df = spark.createDataFrame(schema_data,schema=["column_name","Data_Type","Nullable"])
display(schema_df)
```
