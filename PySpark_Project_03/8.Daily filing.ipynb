### 5.Changing datatype of column

```
srting_to_dateType = ["Payment_datetime", "Payment_date", "cust_first_filing_date", "cust_last_filing_date", "Merchant_first_filing_date", "merchant_last_filing_date","filing_timestamp"]
srting_to_dateType
```

```
for column in srting_to_dateType:
    df2 = df2.withColumn(column,to_date(col(column),"yyyy-mm-dd"))
```

```
df2.select(srting_to_dateType).dtypes
```

```
string_to_int64 = ["Gross_revenue","Net_Revenue"]
string_to_int64
```

```
for column in string_to_int64:
    df2 = df2.withColumn(column,col(column).cast(LongType()))
```

```
df2.select(string_to_int64).dtypes
```

```
display(df2.printSchema())
```

### 6.Handling Null Values

```
df2 = df2.fillna(0,subset=string_to_int64)
```

# Analysis

### 7.Partner wise filing (Using Pie Chart)

```
partner_wise_filing = df2.groupBy("sp_name").agg(count("emitra_id").alias("filing"))
display(partner_wise_filing)
partner_wise_filing = partner_wise_filing.toPandas()

fig = px.pie(
    partner_wise_filing,
    names='sp_name',
    values='filing',
    title="Filing Distrubution By Partner"
)

fig.update_layout(
    title_text = "<b> Filing Distribution By Partner </b>",
    title_x = 0.5
)
fig.update_traces(
    textfont = dict(size=10),
    ##texttemplate = "%{value:.0f}"
)
fig.show()
```

```
df2.printSchema()
```

```
daily_filing = df2.groupBy("Payment_date").agg(count("emitra_id").alias("filing"))
display(daily_filing)
daily_filing = daily_filing.toPandas()
```

```
fig = px.line(
    daily_filing,
    x = 'Payment_date',
    y = 'filing',
    title="Daily Filing Trends",
    labels={"Payment_date": "Payment Date", "filing": "Number of Filings"},
    markers=True,  # Adds data points on the line
    line_shape="linear"  # Ensures a smooth, continuous line
)

fig.update_layout(
    xaxis_title="Payment Date",
    yaxis_title="Number of Filings",
    xaxis=dict(showgrid=True),
    yaxis=dict(showgrid=True),
    template="plotly_white"  # Clean background
)

fig.show()
```
