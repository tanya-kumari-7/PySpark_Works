# Analysis

### 7.Partner wise filing (Using Pie Chart)

```
partner_wise_filing = df2.groupBy("sp_name").agg(count("emitra_id").alias("filing"))
display(partner_wise_filing)
partner_wise_filing = partner_wise_filing.toPandas()

fig = px.pie(
    partner_wise_filing,
    names='sp_name',
    values='filing',
    title="Filing Distrubution By Partner"
)

fig.update_layout(
    title_text = "<b> Filing Distribution By Partner </b>",
    title_x = 0.5
)
fig.update_traces(
    textfont = dict(size=10)
    ##texttemplate = "%{value:.0f}"
)
fig.show()
```

### 9. Filing Distribution By Assesment Year (Using Bar Graph)

```
filing_by_AY = df2.groupBy("assessment_year").agg(
    count("emitra_id").alias("filing"),
    sum("Gross_revenue").alias("Gross_Revenue"),
    sum("Net_revenue").alias("Net_Revenue")
)
filing_by_AY = filing_by_AY.filter(
    (~col("assessment_year").like("%a).%")) &
    (~col("assessment_year").like("%2025%"))
    ) 


filing_by_AY = filing_by_AY.toPandas()
display(filing_by_AY)
```

```
fig = px.bar(
    filing_by_AY,
    x='assessment_year',
    y= 'filing',
    text='filing',
    title='Filing Distribution By AY'
)

fig.update_layout(
    title_x = 0.5,
    plot_bgcolor='white'
)



fig.show()
```

### 10. Filing Distribution with Revenue By Assesment Year (Using Bar Graph)

```
filing_by_AY2 = filing_by_AY.melt(id_vars=['assessment_year'], 
                                       value_vars=['filing', 'Gross_Revenue','Net_Revenue'],
                                       var_name='metric', 
                                       value_name='value')
display(filing_by_AY2)
```

```
fig = px.bar(
    filing_by_AY2,
    x='assessment_year',
    y='value',
    color='metric',  # Different colors for 'filing' and 'Gross_Revenue'
    text='value', 
    barmode='group',  # Group bars side by side,
    title="Filing Distribution By AY"
)
fig.update_layout(
    title_x = 0.5,
    plot_bgcolor = 'white'


)

fig.show()
```


### 11.Solving Date format issue 

```
# Create a new column "payment_month" as the first day of the month
df2 = df2.withColumn("payment_month_m", date_trunc("month", col("Payment_date")))

Monthly_revenue = (df2
    .groupBy("payment_month_m")
    .agg(sum("Net_revenue").alias("Net_Revenue"),count("emitra_id").alias("filing"))
    .orderBy("payment_month_m")  # This is a date column
)

Monthly_revenue.show()

Monthly_revenue = Monthly_revenue.withColumn("payment_month_m",date_format(col("payment_month_m"),"MMM-yy"))
Monthly_revenue = Monthly_revenue.toPandas()
display(Monthly_revenue)
```

### 12.Monthly filing distribution (Using Bar Graph)

```
fig = px.bar(
  Monthly_revenue,
  x="payment_month_m",
  y= "filing",
  text='filing',
  title="Monthly Filing Distribution"
)
fig.update_layout(
  title_x = 0.5,
  plot_bgcolor = "white",
)

fig.show()
```

### 13.Monthly Filing and VLEs Distribution(Using Bar Graph)

```
monthly_filing_by_VLE = df2.groupBy("payment_month_m").agg(
    countDistinct("sso_id").alias("VLEs"),
    count("emitra_id").alias("filing")
    ) .orderBy("payment_month_m") 

monthly_filing_by_VLE=monthly_filing_by_VLE.withColumn("Month",date_format(col("payment_month_m"),"MMM-yy"))
monthly_filing_by_VLE = monthly_filing_by_VLE.toPandas()
monthly_filing_by_VLE = pd.melt(
    monthly_filing_by_VLE,
    id_vars = ["Month"],
    value_vars = ["VLEs","filing"],
    var_name = "Metrics",
    value_name = "Value"
)

display(monthly_filing_by_VLE)
```

```
fig = px.bar(
    monthly_filing_by_VLE,
    x='Month',
    y="Value",
    color="Metrics",
    text="Value",
    title="Monthly Filing and VLEs Distribution",
    barmode="group"


)

fig.update_layout(
    title_x = 0.5,
    plot_bgcolor = 'white'


)

fig.show()
```
