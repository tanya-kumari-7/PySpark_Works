```
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()

from pyspark.sql.functions import col,lit,round,sum,avg,max,min,mean,count,when,udf
import random
from faker import Faker
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
```

```

columns = StructType([
    StructField("StudentName", StringType(), True),
    StructField("Age", IntegerType(), True),
    StructField("Grade", StringType(), True),
    StructField("Gender", StringType(), True),
    StructField("Address", StringType(), True),
    StructField("City", StringType(), True),
    StructField("State", StringType(), True),
    StructField("Country", StringType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone", StringType(), True),
    StructField("Subject", StringType(), True)
])

fake = Faker()
# Generate 200 rows of random student data
data = [
    (
        fake.name(),
        random.randint(18, 25),
        random.choice(["A", "B", "C", "D", "E"]),
        random.choice(["Male", "Female"]),
        fake.address(),
        fake.city(),
        fake.state(),
        fake.country(),
        fake.email(),
        fake.phone_number(),
        random.choice(["Hindi", "Math",'PH'])
    ) for _ in range(200)
]

# Create DataFrame
df = spark.createDataFrame(data, columns)


```

```
# CREATE A NEW COLUMN HAVING VALUE OF  MARKS + Total Marks
def Updated_marks_func(Marks,Total_Marks):
    return Marks + Total_Marks

Updated_Marks=udf(lambda x,y : Updated_marks_func(x,y),IntegerType())


df=df.withColumn("Updated_Marks",lit(10000))
df.select("Marks",'Updated_Marks').show()
df=df.withColumn('Bonus',Updated_Marks(df.Marks,df.Updated_Marks))
df.select('Marks','Updated_Marks','Bonus').show()
```

```
def get_incre(country,Marks,Bonus):
    incre = 0
    if country == 'Korea':
        incre = Marks * 0.10
        incre += Bonus * 0.05
    elif country == 'Guinea':
        incre = Marks * 0.12
        incre += Bonus * 0.05
    return incre

increUDF=udf(lambda x,y,z : get_incre(x,y,z),IntegerType())

df=df.withColumn("Incremental_marks",increUDF(df.Country,df.Marks,df.Bonus))
df.select("Country",'Marks','Bonus','Incremental_marks').where(df.Country == 'Korea').show()
```

```


# Define the function
def get_incre(country, Marks, Bonus):
    increment = 0  # Use a different variable name
    if country == 'Korea':
        increment = Marks * 0.50
        increment += Bonus * 0.60
    elif country == 'Guinea':
        increment = Marks * 0.75
        increment += Bonus * 0.80
    return increment

# Create the UDF
increUDF = udf(lambda x, y, z: get_incre(x, y, z), IntegerType())

# Add the new column
df = df.withColumn("Incremental_marks", increUDF(df.Country, df.Marks, df.Bonus))

# Display the result
df.select("Country", "Marks", "Bonus", "Incremental_marks").filter(df.Country == 'Korea').show()
```
