```
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()

from pyspark.sql.functions import col,lit,round
import random
from faker import Faker
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
```

```

columns = StructType([
    StructField("StudentName", StringType(), True),
    StructField("Age", IntegerType(), True),
    StructField("Grade", StringType(), True),
    StructField("Gender", StringType(), True),
    StructField("Address", StringType(), True),
    StructField("City", StringType(), True),
    StructField("State", StringType(), True),
    StructField("Country", StringType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone", StringType(), True)
])

fake = Faker()
# Generate 200 rows of random student data
data = [
    (
        fake.name(),
        random.randint(18, 25),
        random.choice(["A", "B", "C", "D", "E"]),
        random.choice(["Male", "Female"]),
        fake.address(),
        fake.city(),
        fake.state(),
        fake.country(),
        fake.email(),
        fake.phone_number()
    ) for _ in range(200)
]

# Create DataFrame
df = spark.createDataFrame(data, columns)


```

```
# Filtering using isin function

df.select("Student_name","Gender","Age","Country").filter(df.Country.isin(["Kazakhstan","Spain"])).show(5)
```

```
# filter all data whoes country starts with "K"

df.select("Student_name","Gender","Age","Country").filter(df.Country.startswith("K")).show(5)
```

```
# filter all data whoes country end with "K"

df.select("Student_name","Gender","Age","Country").filter(df.Country.endswith("n")).show(5)
```

```
# filter all data whoes country contains with "An"

df.select("Student_name","Gender","Age","Country").filter(df.Country.contains("An")).show(5)
```

```
# filter all data whoes country starts with "An" using(Like)

df.select("Student_name","Gender","Age","Country").filter(df.Country.like("An%")).show(5)
```
