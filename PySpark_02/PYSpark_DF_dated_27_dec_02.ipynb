```
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()

from pyspark.sql.functions import col,lit,round,sum,avg,max,min,mean,count,when
import random
from faker import Faker
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
```

```

columns = StructType([
    StructField("StudentName", StringType(), True),
    StructField("Age", IntegerType(), True),
    StructField("Grade", StringType(), True),
    StructField("Gender", StringType(), True),
    StructField("Address", StringType(), True),
    StructField("City", StringType(), True),
    StructField("State", StringType(), True),
    StructField("Country", StringType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone", StringType(), True),
    StructField("Subject", StringType(), True)
])

fake = Faker()
# Generate 200 rows of random student data
data = [
    (
        fake.name(),
        random.randint(18, 25),
        random.choice(["A", "B", "C", "D", "E"]),
        random.choice(["Male", "Female"]),
        fake.address(),
        fake.city(),
        fake.state(),
        fake.country(),
        fake.email(),
        fake.phone_number(),
        random.choice(["Hindi", "Math",'PH'])
    ) for _ in range(200)
]

# Create DataFrame
df = spark.createDataFrame(data, columns)


```

```
df = df.withColumn(
    "Age_Group",
    when((df.Age >= 0) & (df.Age <= 30), "0-30 Group")
    .when((df.Age > 30) & (df.Age <= 60), "31-60 Group")
    .otherwise("60+ Group")
)
df.printSchema()
```

```
df.groupBy("Subject","Age_Group")\
    .agg(
        min('Marks').alias("Min_Marks"),
        max('Marks').alias('Max_Marks'),
        avg('Marks').alias('Avg_marks')
    )\
.show()
```
